```python
import os
import geopandas as gpd
```

```python
folder_path = '/content/drive/MyDrive/2026 Spring/Capstone/CAPSTONE WORK PROGRESS/Data/Boundaries/London-wards-2018'
```

```python
file_path = os.path.join(folder_path, 'origdata', 'London-wards-2018_ESRI', 'London_Ward_CityMerged.shp')
```

```python
wards = gpd.read_file(file_path)
city_of_london = wards[wards['DISTRICT'] == 'City and County of the City of London']
city_of_london
```

```python
output_path = os.path.join(folder_path, 'output', 'city_of_london.geojson')
city_of_london.to_file(output_path, driver='GeoJSON')
print(f"Successfully exported to {output_path}")
```
