```python
import os
import geopandas as gpd
```

```python
folder_path = '/green-projects/project-urban_colocation_intelligence/workspace/share/data/boundary/London_LSOA_2021'
data_folder_path = os.path.join(folder_path, 'origdata', 'LB_shp')
```

# **City of London**

```python
file_path = os.path.join(data_folder_path, 'City of London.shp')
city_of_london = gpd.read_file(file_path)
city_of_london
```

```python
output_path = os.path.join(folder_path, 'output', 'city_of_london.geojson')
city_of_london.to_file(output_path, driver='GeoJSON')
print(f"Successfully exported to {output_path}")
```

# **Canary Wharf**

```python
file_path = os.path.join(data_folder_path, 'Tower Hamlets.shp')
tower_hamlets = gpd.read_file(file_path)
tower_hamlets
```

**Canary Wharf's `lsoa21cd`**: E01004277, E01032771, E01032780, E01032782

```python
canary_wharf_lsoa = ['E01004277', 'E01032771', 'E01032780', 'E01032782']
canary_wharf = tower_hamlets[tower_hamlets['lsoa21cd'].isin(canary_wharf_lsoa)]
canary_wharf
```

```python
output_path = os.path.join(folder_path, 'output', 'canary_wharf.geojson')
canary_wharf.to_file(output_path, driver='GeoJSON')
print(f"Successfully exported to {output_path}")
```
